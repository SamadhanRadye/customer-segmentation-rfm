## Step 2: RFM Analysis (Main Goal)

> After the above EDA, we'll build:   

### R = Recency
> How recently did the customer purchase?

### F = Frequency
> How often does the customer purchase?

### M = Monetary
> How much money has the customer spent?

```python
print(df.shape)
print(df["CustomerID"].nunique())
print(df["Country"].nunique())
print(df["InvoiceDate"].min())
print(df["InvoiceDate"].max())
```

```python
snapshot_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

print(snapshot_date)
```

```python
rfm = df.groupby('CustomerID').agg({
    'InvoiceDate': lambda x : (snapshot_date - x.max()).days,
    "InvoiceNo": "nunique",
    'Total Amount': 'sum'
})

rfm.columns = ["Recency", "Frequency", "Monetary"]
rfm.head()
```

```python
rfm.describe().T
```

>  Recency Mean = 92 days   (Average customer hasn't purchased for about 3 months)  
>  Frequency Median = 2     (Most customers only placed 2 orders)                 
>  Frequency Max = 209      (A few customers order very frequently)                 
>  Monetary Median = 668    (Typical customer spends around 668)                   
>  Monetary Max = 280,206   (Some VIP customers spend huge amounts)             


```python
rfm["R_Score"] = pd.qcut(
    rfm["Recency"],
    q=4,
    labels=[4,3,2,1]
)
```

> -  2 days ago    → Score 4
> -  20 days ago   → Score 3
> -  100 days ago  → Score 2
> -  300 days ago  → Score 1

```python
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"),
    q=4,
    labels=[1,2,3,4]
)
```

> -  1 order   → Score 1
> -  20 orders → Score 4

```python
rfm["M_Score"] = pd.qcut(
    rfm["Monetary"],
    q=4,
    labels=[1,2,3,4]
)
```

> -  ₹100    → Score 1
> -  ₹50,000 → Score 4

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)
```

```python
import numpy as np

def remove_outliers(df,col_name):
    Q1,Q3 = np.percentile(df[col_name], [25,75])
    IQR = Q3 - Q1
    minimum = Q1 - 1.5*IQR
    maximum = Q3 + 1.5*IQR
    return df[(df[col_name] >= minimum) & (df[col_name] <= maximum)]
```

```python
rfm = remove_outliers(rfm , 'Frequency')
rfm = remove_outliers(rfm , 'Monetary')
```

-  We usually do not remove outliers from `Recency`, because customers who haven't purchased for a long time are genuinely important for segmentation.

```python
from sklearn.preprocessing import StandardScaler

scale = StandardScaler()
scaled_rfm = scale.fit_transform(rfm[['Recency','Frequency','Monetary']])
```

```python
from sklearn.cluster import KMeans

wcss = []

for i in range(1,21):
    kmeans = KMeans(n_clusters=i)
    kmeans.fit(scaled_rfm)
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(7,5))
plt.plot(range(1,21), wcss, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score

for i in range(2,21):
    kmeans = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(scaled_rfm)

    score = silhouette_score(scaled_rfm, labels)

    print(f"{i} clusters : {score:.3f}")
```

#### The optimal number of clusters was selected as 3 based on the silhouette score, which achieved the highest value (0.453). Although the Elbow Method suggested several possible values, 3 clusters provided the best balance between compactness and separation.

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

rfm["Cluster"] = kmeans.fit_predict(scaled_rfm)
```

```python
centers = pca.transform(kmeans.cluster_centers_)
```

### Regular Customers

-  Encourage repeat purchases through personalized recommendations.
